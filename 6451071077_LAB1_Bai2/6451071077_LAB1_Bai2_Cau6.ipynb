```python
print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

df = pd.read_excel('Online Retail.xlsx')

# 1. Làm sạch tên cột để tránh lỗi khoảng trắng thừa
df.columns = df.columns.str.strip()

# 2. Xử lý cột InvoiceNo và lọc bỏ hóa đơn hủy (có chữ 'C')
df['InvoiceNo'] = df['InvoiceNo'].astype(str)
df = df[~df['InvoiceNo'].str.contains('C', na=False)]

# 3. Tạo biến basket từ dữ liệu United Kingdom
basket = (df[df['Country'] == "United Kingdom"]
          .groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum()
          .unstack()
          .reset_index()
          .fillna(0)
          .set_index('InvoiceNo'))

# 4. Hiển thị 10 dòng đầu tiên của bảng basket
basket.head(10)
```
