```python
print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

transactions = []
with open("groceries.csv", "r", encoding="utf-8") as f:
    for line in f:
        items = [item.strip() for item in line.strip().split(",") if item.strip()]
        if items:
            transactions.append(items)

flat_items = [item for t in transactions for item in t]
item_counts = pd.Series(flat_items).value_counts()

plt.figure(figsize=(12, 8))
item_counts.head(20).sort_values().plot(kind="barh", color="skyblue")
plt.title("Top 20 mặt hàng phổ biến nhất trong dataset Groceries")
plt.xlabel("Tần suất xuất hiện")
plt.ylabel("Mặt hàng")
plt.show()
```
