```python
print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")
import pandas as pd

transactions = []
with open("groceries.csv", "r", encoding="utf-8") as f:
    for line in f:
        items = [item.strip() for item in line.strip().split(",") if item.strip()]
        if items:
            transactions.append(items)

# Lọc bỏ các giao dịch trống hoặc không hợp lệ
valid_transactions = [t for t in transactions if len(t) > 0]
print(f"Số lượng giao dịch hợp lệ sau khi xử lý: {len(valid_transactions)}")
```
