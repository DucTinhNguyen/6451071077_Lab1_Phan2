```python
print("Họ tên: NguyenDucTinh")
print("Mã số sinh viên: 6451071077")

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules

df = pd.read_excel('Online Retail.xlsx')

# 1. Tự động xóa khoảng trắng thừa ở tất cả các tên cột để tránh lỗi KeyError
df.columns = df.columns.str.strip()

# 2. Xử lý cột InvoiceNo (phòng trường hợp có hóa đơn hủy chứa chữ 'C' hoặc kiểu float)
df['InvoiceNo'] = df['InvoiceNo'].astype(str)
df = df[~df['InvoiceNo'].str.contains('C', na=False)]

# 3. Tạo bảng basket (giỏ hàng) cho Germany
basket = (df[df['Country'] == "Germany"]
          .groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum().unstack().reset_index().fillna(0)
          .set_index('InvoiceNo'))

print("Đã tạo xong bảng basket thành công! Kích thước bảng:", basket.shape)
```
